# Inspection of dataset

```python
import numpy as np
import pandas as pd

# 1. Load and inspect the dataset
print("\n==1.load and inspect ===")

df = pd.read_csv("/content/Student_Performance_Dataset_700_Rows (1).csv")

print("STUDENT PERFORMANCE DATASET")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)
```

# missing values and dublicate student Id

```python
# 2. Check missing values and duplicate Student_ID
print("\n==2.check missind value and duplicate student_ID===")

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Student IDs:")
print(df["Student_ID"].duplicated().sum())
```

# dublicate id

```python

# 3. Check missing values and duplicate Student_ID
print("\n==2.check missind value and duplicate student_ID===")

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Student IDs:")
print(df["Student_ID"].duplicated().sum())
```

### Check marks outside 1 to 100

```python
# 4. Check marks outside 0 to 100
print("\n==4.Check marks outside 0 to 100===")

marks_columns = ["Python_Marks", "Maths_Marks", "Physics_Marks"]

print("\nChecking Marks Outside 0-100:")

for col in marks_columns:
    print(col, ":", ((df[col] < 0) | (df[col] > 100)).sum())
```

# calculate total and average marks

```python
# 5. Calculate Total and Average Marks
print("\n==5.Calculate Total and Average Marks===")

df["Total_Marks"] = df[marks_columns].sum(axis=1)

df["Average_Marks"] = df[marks_columns].mean(axis=1)

print("\nTotal and Average Marks:")
print(df[["Name", "Total_Marks", "Average_Marks"]].head())
```

# create grade columns

```python
# 6. Create Grade Column
print("\n==6.Create Grade Column===")

df["Grade"] = pd.cut(
    df["Average_Marks"],
    bins=[-1, 39, 59, 69, 79, 100],
    labels=["F", "D", "C", "B", "A"]
)
```

# Department wise Average marks

```python
# 7. Department-wise Average Marks
print("\n==7.Department-wise Average Marks===")

print("\nDepartment-wise Average Marks:")

department_avg = df.groupby("Department")["Average_Marks"].mean()

print(department_avg)
```

# Find top 10 students

```python
# 8. Find Top 10 Students
print("\n==8.Find Top1 10 students")

print("\nTop 10 Students:")

top_10 = df.sort_values(
    by="Average_Marks",
    ascending=False
).head(10)

print(top_10[["Name", "Department", "Average_Marks", "Grade"]])
```

# Identify students below 40 percent

```python
# 9. Identify Students Below 40% Average Marks
print("\n==9.Identify Students Below 40% Average Marks===")

print("\nStudents Below 40% Average Marks:")

at_risk = df[df["Average_Marks"] < 40]

print(at_risk[["Student_ID", "Name", "Average_Marks"]])

print("\nTotal At-Risk Students:", len(at_risk))

```

# Compare marks based on attendance

```python
# 10. Compare Marks Based on Attendance
print("\n==10. Compare Marks Based on Attendance===")

df["Attendance_Band"] = pd.cut(
    df["Attendance"],
    bins=[-1, 59, 75, 100],
    labels=["Low", "Medium", "High"]
)

print("\nAverage Marks by Attendance Band:")

attendance_analysis = df.groupby(
    "Attendance_Band",
    observed=False
)["Average_Marks"].mean()

print(attendance_analysis)
```

# relationship between study hours and average marks

```python

# 11. Relationship Between Study Hours and Average Marks
print("\n==11.Relationship Between study Hours and AVERAGE Marks===")

print("\nStudy Hours and Average Marks:")

study_analysis = df.groupby("Study_Hours")["Average_Marks"].mean()

print(study_analysis)

print("\nCorrelation Between Study Hours and Average Marks:")

print(df["Study_Hours"].corr(df["Average_Marks"]))

```

# compare students with and without internet access

```python
# 12. Compare Students With and Without Internet Access
print("\n==12. Compare students With Without Internet Access===")


print("\nInternet Access Analysis:")

internet_analysis = df.groupby("Internet_Access")[
    "Average_Marks"
].mean()

print(internet_analysis)
```

# Calculations

```python
# 13. Calculate Median, Standard Deviation and Quartiles
# Using NumPy for Python_Marks
print("/n==13. Calculate Median,Standard Deviation and Quartiles===")

python_marks = df["Python_Marks"]

print("\nPython Marks Statistics:")

print("Median:", np.median(python_marks))

print("Standard Deviation:", np.std(python_marks))

print("First Quartile (Q1):", np.percentile(python_marks, 25))

print("Second Quartile (Q2):", np.percentile(python_marks, 50))

print("Third Quartile (Q3):", np.percentile(python_marks, 75))
```

# Identify score outliers

```python
# 14. Identify Score Outliers Using IQR
print("\n==14. Identify Score Outliers Using IQR===")

print("\nOutliers Using IQR Method:")

for col in marks_columns:

    Q1 = df[col].quantile(0.25)

    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR

    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print("\n", col)

    print("Lower Limit:", lower_limit)

    print("Upper Limit:", upper_limit)

    print("Number of Outliers:", len(outliers))

```
